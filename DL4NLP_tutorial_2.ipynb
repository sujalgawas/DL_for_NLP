```python
import matplotlib.pyplot as plt

def plot(x:list,y:list,label:str):
  plt.figure(figsize=(10,6))
  plt.plot(x,y,label=label,color='blue')
  plt.xlabel("input")
  plt.ylabel("output")
  plt.grid()
  plt.legend()
  plt.show()
```

```python
#sigmoid

def sigmoid(x : list):
  #improvement: use np.exp(-x) instead of writing 2.71828 ** (-x)
  return 1 / (1 + 2.71828 ** (-x))

sig = sigmoid(x)

plot(x,sig,"Sigmoid")
```

```python
#tanh

def tanh(x):
  return (np.exp(x) - np.exp(-x)) / (np.exp(-x) + np.exp(x))

tanh_y = tanh(x)

plot(x,tanh_y,"Tanh")
```

```python
#relu
def ReLU(x):
  #prefered np.maximum(0,x)
  return np.where(x < 0,0,x)

ReLU_y = ReLU(x)

plot(x,ReLU_y,"ReLU")
```

```python
#leaky ReLU
def leaky_ReLU(x,alpa=0.01):
  return np.where(x < 0, x * alpa , x)

leaky_ReLU_y = leaky_ReLU(x)

plot(x,leaky_ReLU_y,"leaky_ReLU")
```

```python
#softmax

def softmax(x):
  exp_x = np.exp(x - np.max(x))
  return exp_x / exp_x.sum(axis=0)

softmax_y = softmax(x)

plot(x,softmax_y,"softmax")
```

```python
#loss_func() and grad()

#loss function : y = x^2
def loss_func(x):
  return x ** 2

#Derivation of loss function : dy/dx = 2 * x
def gradient(x):
  return 2 * x
```

```python
def gradient_descent(start, lr, step):
  x = start
  history = []

  for _ in range(step):
    loss = loss_func(x)
    grad = gradient(x)
    history.append((x,loss))

    x -= lr * grad

  return history
```

```python
history = gradient_descent(10,0.01,100)
```

```python
x_values, loss_values = zip(*history)

# Plotting the Loss Function
x_plot = np.linspace(-10, 10, 500)
y_plot = loss_func(x_plot)

plt.figure(figsize=(10, 6))
plt.plot(x_plot, y_plot, label='Loss Function y = x^2', color='blue')
plt.scatter(x_values, loss_values, color='red', label='Gradient Descent Steps')
plt.title('Gradient Descent Visualization')
plt.xlabel('x')
plt.ylabel('Loss')
plt.legend()
plt.grid()
plt.show()
```
