```python
import requests

url = "https://raw.githubusercontent.com/karpathy/char-rnn/master/data/tinyshakespeare/input.txt"

text = requests.get(url).text
```

```python
n_numbers = 3
inputs = "Once upon a"
```

```python
from collections import Counter

def generate_ngram(text, n):
  #converter text to list for Counter
  ls = list(map(str, text.split()))

  corpus = []
  i = 0
  length = len(ls) - n + 1

  while i < length:
    corpus.append(tuple(ls[i: i + n]))
    i += 1

  ls_counter = Counter(ls)
  text_counter = Counter(corpus)

  return text_counter, ls

model, corpus = generate_ngram(text, n_numbers)

corpus = list(set(corpus))

print(model)
```

```python
def word_pred(input):
    word = None
    prob = -1
    context = tuple(input)

    V = len(corpus)
    denom = 0
    counts = {}
    for x in corpus:
        candidate = context + (x,)
        count = model.get(candidate, 0)
        counts[x] = count
        denom += count

    denom += V
    for x, count in counts.items():
        prob_W = (count + 1) / denom
        if prob_W > prob:
            prob = prob_W
            word = x

    return word
```

```python
def generate_text(model, n, input, max_len):
    input = list(map(str, input.split(" ")))

    for _ in range(len(input), max_len):
        length = len(input)
        context = input[length - (n - 1):length]
        _next_word = word_pred(context)
        if _next_word is None:
            break
        input.append(_next_word)
    return input

output = generate_text(model, n_numbers, inputs, 20)

print(*output)
```
